### 3. Data Aggregation (create new DataFrame according the previous analysis)

```python
# NEED TO CALCULATE THE NUMBER OF LEVEL 2 AND E DC FAST CHARGERS IN STATIONS IN EACH COUNTY

# 1. Fill NaN values in charger count columns with 0 before summing
final_ev_stations['EV Level2 EVSE Num'] = final_ev_stations['EV Level2 EVSE Num'].fillna(0)
final_ev_stations['EV DC Fast Count'] = final_ev_stations['EV DC Fast Count'].fillna(0)

# 2. Group by county and sum the number of Level 2 and DC Fast charging stations, and count the total stations
charging_stations_per_county = final_ev_stations.groupby('county').agg(
    total_level2_chargers=('EV Level2 EVSE Num', 'sum'),
    total_dc_fast_chargers=('EV DC Fast Count', 'sum'),
    num_charging_stations=('ID', 'count') # Count the number of unique station IDs per county
).reset_index()

# 3. Convert counts to integer type as they represent whole numbers
charging_stations_per_county['total_level2_chargers'] = charging_stations_per_county['total_level2_chargers'].astype(int)
charging_stations_per_county['total_dc_fast_chargers'] = charging_stations_per_county['total_dc_fast_chargers'].astype(int)

# 4. Add a new column for total number of chargers
charging_stations_per_county['total_chargers'] = charging_stations_per_county['total_level2_chargers'] + charging_stations_per_county['total_dc_fast_chargers']

display(charging_stations_per_county.head())
```

```python
# 5. Save final file with ev stations
charging_stations_per_county.to_csv('/content/drive/MyDrive/Opt Modeling/Project/charging_stations_per_county.csv', index=False)
```

```python
# look at highway table
highways.head()
```

```python
# find sum in mileage by county
highway_by_county = highways.groupby('County ', as_index=False)["aprox mileage"].sum()
```

```python
# DECIDED TO CONCENTRATE ONLY ON DC FAST CHARGERS PORTS ALLONG IMPORTANT HIGHWAYS

# 1. Deleted unnecessary columns from the charging_stations_per_county table
fastcharging_stations_per_county = charging_stations_per_county.drop(columns=['total_level2_chargers', 'total_chargers'])
fastcharging_stations_per_county.head()
```

```python
# 2. To represent corridor needs, we differentiated charger spacing by county. Front Range counties with higher traffic:
# Adams, Arapahoe, Denver, Douglas, and Boulder, were assigned one fast charger per 3 miles of highway.
# All other counties were assigned one charger per 10 miles

import math
demand = {}

for county, mileage in highway_by_county.values:
  if county in ['Adams', 'Arapahoe', 'Denver', 'Douglas', 'Boulder']:
    divisor = 3.0
  else:
    divisor = 10.0

  demand[county] = math.ceil(mileage / divisor)
```

```python
# 3. Create Demand Dataframe
demand = pd.DataFrame(list(demand.items()), columns=["County", "Demand"])
demand["County"] = demand["County"].str.strip()
```

```python
# 4. Merge Demand Dataframe with fastcharging_stations_per_county
demand = demand.merge(
    fastcharging_stations_per_county[['county', 'total_dc_fast_chargers']],
    left_on='County',
    right_on='county',
    how='inner'
)
```

```python
# 5. Determine how many fast chargers needed in each county
demand['Needed'] = demand['Demand'] - demand['total_dc_fast_chargers']
needed = (demand[demand['Needed'] > 0])[['County', 'Needed']]
needed
```

```python
# 6. Save final file with needed fast chargers by county
needed_fast_chargers = needed
display(needed_fast_chargers.head())
needed_fast_chargers.to_csv('/content/drive/MyDrive/Opt Modeling/Project/needed_fastchargers.csv', index=False)
```

# Section 2

# Optimization Model

### 1. Set up and import Gurobi


```python
!pip install gurobipy

import gurobipy as gp
from gurobipy import GRB
```

### 2. Data Entry

```python
counties = [
    "Archuleta","Bent","Cheyenne","Costilla","Elbert","Gunnison","Huerfano",
    "Lake","Mineral","Ouray","Park","Phillips","Routt","Saguache",
    "San Juan","Teller","Washington","Yuma"
]

needed = {
    "Archuleta":2, "Bent":4, "Cheyenne":3, "Costilla":4, "Elbert":6,
    "Gunnison":3, "Huerfano":5, "Lake":5, "Mineral":2, "Ouray":2,
    "Park":6, "Phillips":3, "Routt":2, "Saguache":4,
    "San Juan":3, "Teller":1, "Washington":5, "Yuma":13
}

# Example regions (replace with your real classification)
regions = {
    "Western": ["Archuleta","Gunnison","Ouray","San Juan","Routt"],
    "Southern": ["Costilla","Huerfano","Saguache"],
    "Plains" : ["Bent","Cheyenne","Phillips","Washington","Yuma"],
    "Central": ["Elbert","Lake","Park","Mineral","Teller"]
}

cost_per_charger = 900000   # $ this is the cost per one charging station with one charger port
budget = 57100000           # $ NEVI program budget


# Region fairness (example: each region must get at least 10% of spend)
min_region_percent = 0.10
```

### 3. Set up the Linear Programming Model

```python
# ----------------------------
# MODEL
# ----------------------------
m = gp.Model("Colorado_EV_Planning")

# DECISION VARIABLES: x[c] = number of fast chargers built in county c
x = m.addVars(counties, vtype=GRB.INTEGER, name="x", lb=0)


# OBJECTIVE FUNCTION: maximize satisfied demand
m.setObjective(gp.quicksum(x[c] for c in counties), GRB.MAXIMIZE)


# ============================
# CONSTRAINT 1: Budget
# ============================

m.addConstr(gp.quicksum(cost_per_charger * x[c]
                        for c in counties) <= budget, "Budget")


# ============================
# CONSTRAINT 2: County minimum and maximum
# (example: at least 0, at most the demand)
# ============================

for c in counties:
    m.addConstr(x[c] <= needed[c], name=f"MaxInCounty[{c}]")
    m.addConstr(x[c] >= 0, name=f"MinInCounty[{c}]")


# ============================
# CONSTRAINT 3: Regional fairness constraints
# Each region must receive at least a fraction of total investment
# ============================

total_spend = gp.quicksum(cost_per_charger * x[c] for c in counties)

for r, rcounties in regions.items():
    m.addConstr(
        gp.quicksum(cost_per_charger * x[c] for c in rcounties)
        >= min_region_percent * total_spend,
        name=f"RegionMinSpend[{r}]"
    )

# ============================
# CONSTRAINT 4: County fairness constraints
# Each county must receive at least one charging station
# ============================

m.addConstrs((x[c] >= 1 for c in counties), name="AtLeastOne")
```

### 4. Results

```python
# Solve
m.optimize()

# Print results
print("\nOptimal number of charging stations with one fast charger to build per county:\n")

print(f"{'County':<15}{'Stations':<10}{'Cost':<15} \n")
for c in counties:
    print(f"{c:<15}{int(x[c].X):<10}${x[c].X * cost_per_charger:,.0f}")

print(f"\nTotal number of charging stations with one fast charger to build: {m.objVal:,.0f}")
print(f"Total cost of building charging stations: $ {m.objVal * cost_per_charger:,.0f}")

# Prepare data for table to save
results_data = []
for c in counties:
    optimized_chargers = x[c].X
    county_demand = needed[c]
    county_total_cost = optimized_chargers * cost_per_charger
    results_data.append({
        "County": c,
        "Demand": county_demand,
        "Optimized Chargers": optimized_chargers,
        "Total Cost": county_total_cost
    })

results_df = pd.DataFrame(results_data)
```

```python
# Save Results DF
results_df.to_csv('/content/drive/MyDrive/Opt Modeling/Project/results_opt.csv', index=False)
```
